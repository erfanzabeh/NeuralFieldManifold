# Inspect per-session EEG sleep decoding results

Run the conversion and analysis scripts from the parent folder first, then use this notebook to browse tables and plots.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Image

ROOT = Path('..').resolve()
tables = ROOT / 'tables'
plots = ROOT / 'plots'

scores = pd.read_csv(tables / 'session_decode_scores.csv').rename(columns={'macro_f1': 'f1'})
counts = pd.read_csv(tables / 'session_class_counts.csv')
display(counts)
display(scores.head(20))
```

```python
f1_pivot = scores.pivot_table(index='session_id', columns='feature_set', values='f1', aggfunc='first')
accuracy_pivot = scores.pivot_table(index='session_id', columns='feature_set', values='accuracy', aggfunc='first')
display(f1_pivot)
display(accuracy_pivot)
summary_paths = [
    plots / 'summary' / 'session_f1_heatmap.png',
    plots / 'summary' / 'session_f1_barplot.png',
    plots / 'summary' / 'session_accuracy_heatmap.png',
    plots / 'summary' / 'session_accuracy_barplot.png',
]
for summary_path in summary_paths:
    if summary_path.exists():
        display(Image(filename=str(summary_path)))
```
