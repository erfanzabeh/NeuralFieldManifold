# Macaque Motor-Cortex LFP Reaching Analysis

This notebook inspects the rebuttal analysis for the Confais/Kilavik/Riehle macaque motor-cortex LFP dataset. The code uses the documented six reach directions and short/long delay labels; it does not include the unrelated grid-cell discussion.

The reviewer-facing torus result is `torus_nonlinear_15`, a slower nonlinear elliptical-torus refit per trial. The earlier `torus_geometry_15` output is kept as a quick screening estimate.

The current reviewer-facing rerun selects tau and embedding dimension separately for each unique LFP trace from signal structure only, then regenerates the decoding figures with those per-trace parameters.
